# PyTorch Basics

PyTorch is a very popular Deep Learning framework. It was originally developed at Meta, but now maintained under Linux foundation. It is an open source framework and widely used in both research and industry for AI projects. It is used for both creation (training) of new models as well as fine-tuning existing models.

## Environment

These are the environment suggestions for a beginning deep learning project using PyTorch. PyTorch is a Python framework. So, a supported version of Python for a given PyTorch version needs to be installed. A separate environment (for example, a kernel for Jupyter Notebooks) would work best. 

### Environment with Python

For example, creating a conda environment with Python version 3.13, use:

```bash
conda create -n pytorch-learning python=3.13
```

### Install packages

Once an environment is created, activate that environment (here named `pytorch-learning`) and install the basic Python packages for a deep learning project. More packages can be installed as needed.

```bash
conda activate pytorch-learning
python -m pip install torch torchvision numpy matplotlib ipykernel
```

Here `conda` is used for managing the Python environment (installing Python, activating the environment); and `pip` is used for installing the packages within the Python environment (pip is part of a Python installation; conda is outside that installation).

Basic deep learning packages are installed here. `torch` is PyTorch. `torchvision` is companion package to PyTorch for computer vision. It also contains well known datasets like MNIST (70k images of handwritten digits 0-9). `numpy` is package for numerical computing. It is not strictly needed for very basic deep learning projects as PyTorch uses tensors and numpy uses arrays. However, numpy is used a lot outside of PyTorch (for example, in pandas, matplotlib, etc.), so, it is installed here. `matplotlib` is for plotting and visualization. `ipykernel` is for Jupyter Notebooks.

## Benefits of PyTorch

### Pythonic

The biggest benefit of PyTorch is that it is 'pythonic'. It fits well within a Python program. The deep learning projects developed with PyTorch look and feel like Python programs.

### Dynamic Computational Graph

Dynamic Computational Graph vs. Static Computational Graph. PyTorch builds the graph dynamically during the forward pass where loss is calculated for a given weight and bias. The graph records all the differentiable operations (squaring, multiplication, addition, etc.) during the forward pass. This graph is used to calculate the gradients (during the optimizer step) and adjust the values of weight and bias during the back propagation step. Weight and bias are tensors with requires_grad set to True.

Then this graph is thrown away and rebuilt during the next cycle of forward pass and the weights and biases are adjusted again during the next back propagation. This dynamic nature helps debugging easy and helps with researching. In earlier frameworks like TensorFlow v1.0, the computational graph is static - built separately and 'frozen'. During each batch of adjusting weights and biases to produce more accurate output, these numbers are fed into the static graph. Debugging becomes difficult and not as friendly for research as PyTorch.

### Ecosystem

PyTorch also has a large ecosystem. Lot of libraries, models, courses, community, and research base. Vast majority of open-weight models (such as Llama, Mistral, Qwen, DeepSeek, and Stable Diffusion) are developed, pre-trained, and released natively in PyTorch. So, development of these models involve using various PyTorch modules like torch.nn, torch.optim, torch.compile, etc.

## Stages in an ML/DL Project

A typical Machine Learning / Deep Learning project goes through multiple stages from beginning (inception) the the end (production). These are:

- Stage 0: Creation of a PRD (idea, problem statement, ML brief, design doc, etc.)
- Stage 1: Data Ingestion
- Stage 2: Data Preparation
- Stage 3: Model Architecture
- Stage 4: Model Training
- Stage 5: Eval and Error Analysis
- Stage 6: Production (Deployment)
- Stage 7: Monitoring and Retraining

### Stage 0: Creation of a PRD

PRD is a product requirements document. Depending on the project scope and company, there could be an ML Brief or a full-fledged design document. A small project could be a one-pager on what the problem is how the proposed project aims to solve it.

A more comprehensive approach is suggested by Google, which includes creation of a Design Document. In this Ideation and Planning exercise, you would start with defining the problem and evaluating whether ML is the right strategy. Then see if the problem has already been solved via the existing models (perhaps an open source model). For a lot of corporate projects, creating a new model from scratch might not be needed. Take an existing model and fine-tune it with the company data.

As a part of this planning exercise, also consider whether enough and good data is available, what is the training cost and the inference cost while it is in production. Plus features, tools, measurements of success need to be thought through. This could result in large or small design document based on the size and complexit of the project.

Google's recommendations on Machine Learning projects is here: [Managing ML Projects](https://developers.google.com/machine-learning/managing-ml-projects)

### Stage 1: Data Ingestion

During this stage, data needed for the ML project is collected. There are two places where the data is obtained from:

- Public / Open Datasets
- Private

**Public / Open Datasets**

Large number of datasets are availably publicly that are free to download. A large number of datasets are available for variety of areas like computer vision, NLP/LLM/audio, Tabular data, Sceintific data, custom data from public APIs, perhaps web scraping. Some popular sites also host a lot of datasets:

- torchvision
- Hugging Face
- Kaggle

Generally these public datasets ('open datasets') are very clean, nicely labeled. However, these datasets are probably more useful for hobby projects or research. In addition, government websites have lot of datasets on economy, employment, demographics, taxes, etc. For a company related task, some of them might be a starting point, but for private tasks, the internal data is needed.

**Private**

This is the data from a company's internal data sources. They become core part of the machine learning solutions that are to be developed in the context of a particular company. This data comes from:

- databases and data warehouses (like PostgreSQL, MySQL, Oracle, Snowflake)
- internal documents (CSV, Excel, PDF, etc.)
- documents stored on S3
- enterprise systems (like Salesforce)
- internal and third-party APIs

This data is generally very dirty and unusuable without further preparation. They will not be the formats that are readily accepted by deep learning frameworks like PyTorch. Most of the time these are complete opposite (in terms of cleanliness and usability) of public datasets used in hobby/research projects.
