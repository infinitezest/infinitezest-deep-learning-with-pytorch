# Tensors

Tensors are the most fundamental data structures of PyTorch. Data that corresponds to inputs (features), outputs (targets), parameters (weights and biases), and various intermediate data representations are stored in tensors. Tensors are multidimensional arrays similar to the ndarrays in NumPy, with a couple of big capabilities. Tensors operations are optimized to execute fast on GPUs. And tensors provide functionality to calculate gradients (the differentiation), which is indispensable for arriving at proper values for parameters (weights and biases).

NumPy (with `ndarray` to represent multidimensional arrays) is a very popular data science package and is used to represent data for packages like pandas and matplotlib. Standard numpy, pandas, matplotlib work on CPUs, while tensors work on CPUs and GPUs (but GPUs are essential for speed for any decent sized deep learning project). PyTorch's tensors can be moved to a different GPU device from CPUs. These vendor/stacks include: Nvidia's CUDA, AMD's ROCm, Apple's MPS, Google's TPU (xla), Intel's training accelerator (hpu), Meta's training/inference ASIC (mtia), etc.

```python
import torch
import numpy as np
import pandas as pd
```

# 1. Tensor Creation

Easiest way to create tensors is from a Python list of numbers. Pass a list as a parameter to tensor creation factory function `torch.tensor`. Actual class for tensors is `Tensor`, but for creation, `torch.tensor()` is used. Two main attributes of a torch.Tensor class are `dtype` which is data type of all the items in the tensor and `shape` tells about the shape of the data (e.g. rows/cols for 2-d data). A tensor holds elements of just one datatype, for example, float32, int64, etc.

```python
# A simple Python list
simple_list = [2, 4, 6]
print(f"Type (class) of the input: {type(simple_list)}")

# Create a tensor 
x = torch.tensor(simple_list)
print(f"Type (class) of the tensor: {type(x)}")
print(f"Data type of numbers in the tensor: {x.dtype}")
```

Tensors can be created from various Python objects holding numeric or boolean data:

- scalars
- list with boolean values
- tuples
- lists with numbers

```python
a_scalar = 3.0
y_s = torch.tensor(a_scalar)
print(f"Tensor from num: {y_s} \ndtype={y_s.dtype} \nshape={y_s.shape}")

bool_list = [True, False, True]
y_b = torch.tensor(bool_list)
print(f"\nTensor from bool list: {y_b}, \ndtype: {y_b.dtype} \nshape: {y_b.shape}")

simple_tuple = (5.6, 8.9)
print(f"\nType (class) of the input: {type(simple_tuple)}")

y = torch.tensor(simple_tuple)
print(f"Tensor from tuple: {y} \ndtype: {y.dtype} \nshape: {y.shape}")

py_list = [12, 3.0, 0.01]
y_t = torch.tensor(py_list)
print(f"\nTensor from different types of numbers: {y_t} \ndtype: {y_t.dtype} \nshape: {y_t.shape}")

print("\nTensor can't hold a String or non-numeric objects:")
py_mixed_list = [4, "A Str", 1.2]
try:
    y_m = torch.tensor(py_mixed_list)
except TypeError:
    print("Invalid data type")
```

## 1.1 Tensor creation from NumPy arrays

This is the most popular way creating tensors - from NumPy arrays. NumPy already provides the functionality for storing data in popular packages like pandas, matplotlib, etc. Preparation (and display, etc.) of data is done in these packages outside PyTorch and for the machine learning purposes (training/inference/etc.), these numpy arrays are converted to tensors from simple functions like `torch.from_numpy` or `torch.tensor`.

- `torch.tensor`: creates a new tensor object by copying the data from NumPy's ndarray. These two objects are independent. And the tensor can be created on CPU or directly on an available GPU (for example, "mps" for Apple's devices like MacBook Pro, or "cuda" for devices with Nvidia GPUs).

- `torch.from_numpy`: Here the created tensor points to the current NumPy ndarray storage. The new tensor does not copy the data. Hence it is on CPU. However, once we have a tensor, it can be moved to a device by using .to("<device_type>") method.

```python
# 2 pixels in a NumPy array.
# Each pixel has 3 channels: RGB
# Each value goes from 0-255
# Hence, the dtype is uint8
numpy_array_pixels = np.array([
    [125, 32, 160],
    [240, 126, 38],
], dtype=np.uint8)

print("\nNumPy Array:")
print("------------")
print(f"Class of NumPy array: {type(numpy_array_pixels)}")
print(f"NumPy array of Pixels (RGB): \n{numpy_array_pixels}")
print(f"Shape: {numpy_array_pixels.shape}")
print(f"Size: {numpy_array_pixels.size}")

# A tensor can be created from the NumPy ndarray
# This tensor is created on MacBookPro, hence the device is mps.
x_pixels = torch.tensor(numpy_array_pixels, device="mps")
print("\nTensor with torch.tensor():")
print("---------------------------")
print(f"Class of Tensor: {type(x_pixels)}")
print(f"Device: {x_pixels.device}")
print(f"Tensor: {x_pixels}")
print(f"Shape: {x_pixels.shape}")
print(f"Sizes at Dim0 - row and Dim1 - col: {x_pixels.size(0)}, {x_pixels.size(1)}")

# Tensor from from_numpy
x_pixels2 = torch.from_numpy(numpy_array_pixels)
print("\nTensor with torch.from_numpy():")
print("-------------------------------")
print(f"Tensor: {x_pixels2}")
print(f"Device: {x_pixels2.device}")

# Moving Tensor to device
x_pixels2 = x_pixels2.to("mps")
print("\nMove Tensor to device:")
print("----------------------")
print(f"Device: {x_pixels2.device}")
```

## 1.2 Tensor Creation from Pandas Dataframe

Pandas is a popular data science package. The main object in pandas is a Data Frame. This is a table like structure (providing functionality somewhat similar to Excel). Pandas can read data from sources like .csv files and from there it can be analyzed, manipulated, and perhaps prepared for training a model. The data frame has an attribute named `values` that gives the entire data behind that data frame as a NumPy ndarray. This can be used to create a tensor either via torch.from_numpy() or torch.tensor() which creates a tensor with copied data, which can be directly created on a GPU.

### Example .csv file

Here the `classification.csv` contains a mapping between a category and an image_num, i.e. to which category an image_num belongs. Here both categories and image_num are in numbers. Separately categories will have textual names (say, banana, apple, etc.) and image_num points to a location of the image (row in some other csv file or name of the image file without extension, etc.)

```python
# Pandas Data Frame.
df = pd.read_csv('./classification.csv')

print("\nData Frame from Pandas:")
print("-----------------------")
print(f"Data Frame: \n {df}")
print(f"\nShape of df: {df.shape}")
print(f"\n3 rows at the Head of df: \n {df.head(3)}")

print(f"\nColumns of df: {df.columns}")
print(f"\nClass of df.values: {type(df.values)}, datatype of df.values: {df.values.dtype}")
print(f"Class of one column: {type(df['category'])}")
```

```python
# Tensors from data frame
tensor_from_df = torch.from_numpy(df.values)
tensor_from_df2 = torch.tensor(df.values, device="mps")
tensor_from_df3 = torch.tensor(df[["category", "image_num"]].to_numpy(), dtype=torch.float32)

print("Pandas Data Frame to PyTorch Tensor:")
print("------------------------------------")
print(f"\nTensor from_numpy: \n{tensor_from_df} \ndtype={tensor_from_df.dtype}")
print(f"\nTensor torch.tensor: \n{tensor_from_df2} \ndtype={tensor_from_df2.dtype}")
print(f"\nTensor with explicit data type: \n{tensor_from_df3} \ndtype: {tensor_from_df3.dtype}")
```

# 2. Tensors with Placeholder Values

You can create tensors of certain shape filled with placeholder values. The popular ones are:

- tensors filled with zeros: torch.zeros
- tensors filled with ones: torch.ones
- tensors filled with random values: torch.rand, torch.randint, etc.
- empty tensors (with unintialized values)
- values in a range
- identity tensor (1s diagonally)

```python
zeros = torch.zeros(2, 3)
print(f"Tensor with zeros: \n{zeros}")

ones = torch.ones(2, 3)
print(f"\nTensor with ones: \n{ones}")

random = torch.rand(2, 3)
print(f"\nRandom values: \n{random}")

empty = torch.empty(2, 3)
print(f"\nEmpty Tensor (with uninitialize values): \n{empty}")

range_tensor = torch.arange(0, 10, step=2)
print(f"\nRange of values: \n{range_tensor}")

linspace_tensor = torch.linspace(0, 1, steps=5)
print(f"\nLinearly spaced: \n{linspace_tensor}")

identity_tensor = torch.eye(3)
print(f"\nIdentity: \n{identity_tensor}")
```

# 3. Shape and Manipulating

Shape of tensor is size along each of the dimensions. A 2-D tensor (a table that has columns and rows) has a shape of (m,n) - for example (2,3). Meaning this tensor has 2 rows and 3 columns. So, this 2-D tensor has 2 dimensions: 0 and 1. So, the dimension 0 would be rows and dimension 1 would be columns. If this 2-D tensor of shape (2,3) is storing the RGB values of pixels, it will have 2 samples (i.e. number along dimension 0 or number of rows). And it will have 3 columns, i.e. the number along dimension 1, presumably each column is for each of RGB (Red, Green, and Blue). Shape can be obtained from `.shape` attribute. You can also use `.size(<dim>)` function to get the size of a particular dimension (in this case .size(0) or .size(1)).

However, batches of images typically have this shape: (N, H, W, C) or (N, C, H, W). For example a shape of (64, 3, 28, 28) would mean the following:

N: 64 samples (images) in a batch
C: 3 channels (RGB - each with values in 0-255)
HxW: 28x28 - the image size is 28 pixels height and 28 pixels width

NumPy and TensorFlow use (N, H, W, C) shape; whereas PyTorch (nn.Conv2d) uses (N, C, H, W) shapes. Shapes can be changed with .permute function.

```python
# Shape and Size of a tensor
x = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])

print("Shape and Sizes a tensor:")
print("------------------------")
print(f"Tensor: \n{x}")

#Shape
print(f"\nShape: {x.shape}")
#Size
print(f"Size of Dimension 0: {x.size(0)}")
print(f"Size of dimension 1: {x.size(1)}")
```

## 3.1 Squeeze and Unsqueeze

.squeeze(<dimension>) removes the dimension if it is a 1, otherwise it leaves it alone. .squeeze() removes all the dimensions with values 1. So, a tensor with shape (1, 3, 4) with .squeeze() explicitly .squeeze(0) with squeezing on dimension 0 will result in a tensor of (3, 4).

.unsqueeze(<dimension>) will add an extra dimension with value 1 at a specified place. If you have one image with shape (3, 28, 28) but a layer is expecting also a batch size, you can use .unsqueeze(0) to get a shape of (1, 3, 28, 28).

```python
print("Unsqueeze: .unsqueeze()")
print("-----------------------")

print(f"\nOriginal tensor: \n{x} \nshape: {x.shape}")

unsqueeze_at_0 = x.unsqueeze(0)
print(f"\nUnsqueezed tensor at Dimension 0: \n{unsqueeze_at_0}")
print(f"Unsqueezed0 shape: {unsqueeze_at_0.shape}")

unsqueeze_at_1 = x.unsqueeze(1)
print(f"\nUnsqueezed tensor at Dimension 1:{unsqueeze_at_1}")
print(f"Unsqueezed1 shape: {unsqueeze_at_1.shape}")
```

```python
# Remove the dimension added to unsqueeze_at_0
print("Squeeze: .squeeze()")
print("-------------------")

squeeze_at_0 = unsqueeze_at_0.squeeze(0)
print(f"\nSqueeze at 0: {squeeze_at_0}")
print(f"Shape: {squeeze_at_0.shape}")
```

## 3.2 Reshape, transpose, and permute

**`.reshape()`** reshapes tensor to the new dimensions. If tensor shape is (2, 3, 4), this tensor has 2x3x4=24 elements. As long as all the elements are used, the tensor can be reshaped to any shape. All these reshapes work: <old tensor>.reshape(2, 6, 2), <old_tensor>.reshape(6, 4) or <old_tensor>.reshape(24, 1) since they are redistributing all the elements across different dimensions.

**`.transpose()`** transposes just two dimensions. If a tensor is (2, 3, 4), a .transpose(0, 1) switches the dimensions 0 and 1.

**`.permute()`** rearranges the dimensions. If a tensor shape is (2, 3, 4), applying .permute(2, 1, 0) will put the dimensions in this order: 2, 1, 0 (in this case reverse). Here 0, 1, 2 are dimension numbers, not the sizes.

```python
print("Reshape: .reshape()")
print("-------------------")

t_r = torch.rand(2, 3, 4)
print(f"\nOriginal: {t_r} \nshape: {t_r.shape}")

t_rs = t_r.reshape(4,6)
print(f"\nReshaped: {t_rs} \nshape: {t_rs.shape}")
```

```python
print("Transpose: .transpose()")
print("-----------------------")

t_pose = torch.rand(2, 3, 4)
print(f"\nOriginal: {t_pose} \nshape: {t_pose.shape}")

t_posed = t_pose.transpose(0, 1)
print(f"\nTransposed (0,1): {t_posed} \nshape: {t_posed.shape}")
```

```python
print("Permute: .permute()")
print("-------------------")

t_perm = torch.rand(2, 3, 4)
print(f"\nOriginal: {t_perm} \nshape: {t_perm.shape}")

t_permed = t_perm.permute(2, 1, 0)
print(f"\nPermuted: {t_permed} \nshape: {t_permed.shape}")
```

## 3.3 Concatenation

`torch.cat()` concatenates the tensors along the specified dimension. If you concatenate two tensors with shape (2, 3) along dimension 0, the resulting tensor will have a shape of (4, 3). If these tensors are concatenated along dimension 1, then they will have a shape of (2, 6).

```python
print("Concatenation: torch.cat()")
print("--------------------------")

t_cat_a = torch.rand(2, 3)
print(f"\ncat_a: {t_cat_a} \nshape: {t_cat_a.shape}")

t_cat_b = torch.rand(2, 3)
print(f"\ncat_b: {t_cat_b} \nshape: {t_cat_b.shape}")

t_cat_0 = torch.cat((t_cat_a, t_cat_b), dim=0)
print(f"\ncat along Dim0: {t_cat_0} \nshape: {t_cat_0.shape}")

t_cat_1 = torch.cat((t_cat_a, t_cat_b), dim=1)
print(f"\n cat along Dim1: {t_cat_1} \nshape: {t_cat_1.shape}")
```

# 4. Indexing

Indexing in tensors works in a familiar fashion for multi-dimensional arrays. Let's say a tensor `t` is shaped (2, 3, 4). Say, it contains 2 gray-scale images of size (3x4).

t[0][0][0] will be the first element (the first pixel in the first row of the first image). And the last element (last pixel in the last row of the 2nd image) can be accessed via either t[-1][-1][-1] or t[1][2][3].

Entire second image can be accessed via t[1].

```python
print("Indexing")
print("--------")

t = torch.rand(2, 3, 4)
print(f"\nTwo full images: \n{t} \nshape: {t.shape}")

print(f"\nFirst pixel in the first row of first image: {t[0][0][0]}")

print(f"\nLast pixel in the last row of second image: {t[-1][-1][-1]}")
print(f"Last pixel in the last row of second image: {t[1][2][3]}")

print(f"\nComplete second image: \n{t[1]}")
```

# 5. Slicing

Slicing follows this format: [start:end:step]. With this format you can keep slicing until the indexes run out.

For example, take a batch of 2 sample images in a tensor named s with this shape: (2, 3, 4). This has 2 images, each 3x4 size.

You can get the second image (with the index 1) with: s[1] - that gives an image of shape (3, 4)
Of that you can get rows 2 and 3 (row index 1 and 2): s[1][1:3]; this is same as s[1][1:3, :] - in the second slicing you are including all the columns.
Of that you can get last two columns: s[1][1:3][:, 2:4]

```python
print("Slicing")
print("-------")

s = torch.rand(2, 3, 4)
print(f"\n2 images 3x4 each: \n{s} \nshape: {s.shape}")

print(f"\n2nd image: \n{s[1]} \nshape: {s[1].shape}")

print(f"\n2nd+3rd rows of the 2nd Image: \n{s[1][1:3]} \nshape: {s[1][1:3].shape}")

print(f"\n3rd+4th columns of 2nd+3rd rows of 2nd Image: \n{s[1][1:3][:, 2:4]} \nshape: {s[1][1:3][:, 2:4].shape}")
```

## 5.1 Accessing Single Item

If there is only one item in the tensor, you can access it via .item(). If there are multiple elements, then use the indexing to access that item. .item() wouldn't work if there are more than 1 item in the tensor.

```python
print("Scalar item via .item()")
print("-----------------------")

i = torch.rand(2, 3, 4)
print(f"\n2 image batch: \n{i} \nshape: {i.shape}")

first_item_tensor = i[0][0][0]
print(f"\nfirst item tensor: {first_item_tensor} \nshape: {first_item_tensor.shape}")

first_item_scalar = first_item_tensor.item()
print(f"\nfirst item scalar: {first_item_scalar}")
```

## 5.2 Indexing with mask

You can create mask (like `mask = y > 0.3`) and apply it to the tensor (like `y[mask]`) get the values matching the mask. mask is a tensor that contains True False values resulting from the y > 0.3 operation on each of the elements in the original tensor. With y[mask] each element is selected only when the mask element is True. Here the boolean indexing is used.

```python
print(f"Indexing with mask")
print(f"------------------")

x_boo = torch.rand(2, 3)
print(f"\nOriginal Tensor: \n{x_boo} \nshape: {x_boo.shape}")

mask = x_boo > 0.6
print(f"\nThe mask tensor with bools that satisfy the condition (> 0.6): \n{mask} \nshape: {mask.shape}")

print(f"\nThe mask applied to the original: \n{x_boo[mask]}")
```

# 6. Math Operations

Addition, multiplication, etc. happen element-wise. Dot product is element wise multiplication and a sum (torch.matmul)

```python
x = torch.rand(3, )
y = torch.rand(3, )

print(f"Tensor 1: {x}")
print(f"Tensor 2: {y}")

t_sum = x + y
print(f"\nElement-wise Sum: {t_sum}")

t_mult = x * y
print(f"Element-wise mult: {t_mult}")

t_matmul = torch.matmul(x, y)
t_hand_calc = x[0]*y[0] + x[1]*y[1] + x[2]*y[2]
t_mult_sum = t_mult.sum()

print(f"\nDot product via matmul: {t_matmul}")
print(f"Dot product hand calculated: {t_hand_calc}")
print(f"Dot product via .sum() on element-wise mult: {t_mult_sum}")
```
